```python
# ==========================================
# Experiment 6: Sentiment Analysis +
# Zero-Shot Document Classification
# Using Hugging Face Transformers
# Google Colab Compatible Version
# ==========================================

!pip -q install transformers torch

from transformers import pipeline

print("Loading models...")

# ------------------------------------------
# 1. Sentiment Analysis
# ------------------------------------------

sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("Sentiment Analysis Model Loaded!")


reviews = [
    "The new smartphone has an amazing camera and battery life!",
    "The delivery was late and the packaging was damaged."
]


print("\n==============================")
print("SENTIMENT ANALYSIS")
print("==============================")


for review in reviews:

    result = sentiment_analyzer(review)[0]

    print("\nReview:", review)
    print(
        "Sentiment:",
        result["label"],
        "| Confidence:",
        round(result["score"], 3)
    )



# ------------------------------------------
# 2. Zero-Shot Document Classification
# ------------------------------------------

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

print("\nZero-Shot Classification Model Loaded!")


document = """
The central bank raised interest rates to control
rising inflation and stabilize the economy.
"""


candidate_labels = [
    "Politics",
    "Economy",
    "Sports",
    "Technology"
]


classification = classifier(
    document,
    candidate_labels
)


print("\n==============================")
print("ZERO-SHOT DOCUMENT CLASSIFICATION")
print("==============================")


print("Document:", document)


for label, score in zip(
    classification["labels"],
    classification["scores"]
):
    print(
        label,
        ":",
        round(score, 3)
    )
```
